```python
%%sql 
CREATE TABLE Enrolled (
  studid CAHR(20),
  cid CHAR(20),
  grade CHAR(10),
  PRIMARY KEY (studid, cid),
  FOREIGN KEY (studid) REFERENCES Students (sid)
    ON DELETE SET NULL
    ON UPDATE CASCADE 
);
```
